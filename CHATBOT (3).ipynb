# SIMPLE REFLEX MODEL

```python
def simple_reflex_model():
    print("--------- SIMPLE REFLEX MODEL---------------")
    print("write bye to exit")
    while(True):
        user_input = input("You:").lower().strip()
        words = user_input.split()
        if "bye" in words:
            print("Bot: Goodbye!")
            break
        elif "hi" in words:
            print("Bot: Hi, How may i help u today!")
        elif "ai" in words:
            print("Bot: Artificial intelligence is the new technology that is advancing at an exponential rate and is helping humans in their daily lives.")
        else:
            print("Bot: I do not understand!!")
        
simple_reflex_model()    
```

# Model Based Chatbot

```python
def performance(response):
    if response == "I do not understand!":
        return "Bad response"
    return "Good response"

def environment():
    return "Jupyter Notebook"

def actuator(response):
    print("Bot:", response)

def sensors():
    return input("You: ").lower().strip()


def model_based_chatbot():
    print("----- MODEL BASED CHATBOT -----")
    print("Type bye to exit")
    print("Environment:", environment())

    state = {"name": None, 
             "topic": None}

    while True:
        user = sensors()

        if user == "bye":
            response = "Goodbye!"

        elif user == "hi":
            response = "Hello! How can I help you?"

        elif user.startswith("my name is "):
            state["name"] = user.replace("my name is ", "", 1)
            response = "Hello " + state["name"]

        elif user == "what is my name":
            response = "Your name is " + str(state["name"])

        elif "ai" in user.split():
            state["topic"] = "Artificial Intelligence"
            response = "AI is a technology that helps humans."

        elif user == "what were we talking about":
            response = "We were talking about " + str(state["topic"])

        else:
            response = "I do not understand!"

        actuator(response)
        print("Performance:", performance(response))

        if user == "bye":
            break


model_based_chatbot()
```
