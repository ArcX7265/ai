# PR4: Vehicle Fault Diagnosis
**CO1, CO2**

Fixed observations, six goals and six rules. No libraries are required. Rule 3 uses OR as previously requested; it is an adaptation of the original case study’s AND condition.

## 1. Observations, goals and rules

```python
observations = ["engine_not_starting", "battery_low", "headlights_dim",
            "fuel_low", "engine_overheating", "coolant_low", "engine_noise",
            "oil_low", "warning_light_on", "starter_clicking", "vehicle_old"]

goals = ["starting_system_fault", "battery_replacement_required",
         "cooling_system_fault", "severe_engine_risk",
         "immediate_service_required", "vehicle_service_required"]

rules = [
    (1, ["battery_low", "headlights_dim", "starter_clicking", "engine_not_starting"], "AND", "starting_system_fault"),
    (2, ["starting_system_fault", "vehicle_old"], "AND", "battery_replacement_required"),
    (3, ["engine_overheating", "coolant_low", "warning_light_on"], "OR", "cooling_system_fault"),
    (4, ["cooling_system_fault", "engine_overheating", "oil_low", "engine_noise"], "AND", "severe_engine_risk"),
    (5, ["severe_engine_risk", "vehicle_old"], "AND", "immediate_service_required"),
    (6, ["starting_system_fault", "immediate_service_required"], "AND", "vehicle_service_required")
]
```

## 2. Forward and backward chaining
The inference engines automatically choose applicable rules from the fixed knowledge base. `all()` means AND and `any()` means OR.

```python
def forward(facts, goal):
    known = set(facts)
    while True:
        before = known.copy()
        for number, conditions, operator, result in rules:
            checks = [x in known for x in conditions]
            satisfied = all(checks) if operator == "AND" else any(checks)
            if satisfied:
                known.add(result)
        if known == before:
            return goal in known


def backward(facts, goal):
    if goal in facts:
        return True
    for number, conditions, operator, result in rules:
        if result == goal:
            checks = [backward(facts, x) for x in conditions]
            satisfied = all(checks) if operator == "AND" else any(checks)
            if satisfied:
                return True
    return False
```

## 3. Select observations and a goal
Enter observation numbers separated by spaces, then choose one goal number. Use valid numbers from the displayed lists. Both methods check the same selected goal.

**Example:** observations `1 2 3 10 11`, goal `2`. Both methods return True. Rules remain fixed; the user selects the known observations and the goal to prove.

```python
for i, observation in enumerate(observations, 1):
    print(i, observation)

numbers = input("Enter observation numbers separated by spaces: ").split()
facts = {observations[int(n) - 1] for n in numbers}

print("\nAvailable goals:")
for i, goal in enumerate(goals, 1):
    print(i, goal)

number = int(input("Enter goal number: "))
goal = goals[number - 1]

print("\nGoal:", goal)
print("Forward chaining:", forward(facts, goal))
print("Backward chaining:", backward(facts, goal))
```
